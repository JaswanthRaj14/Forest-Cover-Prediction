```python
from sklearn.model_selection import train_test_split

X_new = df.drop(columns=["Id", "Cover_Type"])
y_new = df["Cover_Type"]

X_train_new, X_test_new, y_train_new, y_test_new = train_test_split(
    X_new,
    y_new,
    test_size=0.2,
    random_state=2026,
    stratify=y_new
)

# Remove the two constant soil features
constant_cols = ["Soil_Type7", "Soil_Type15"]

X_train_new = X_train_new.drop(columns=constant_cols)
X_test_new = X_test_new.drop(columns=constant_cols)

print("New training shape:", X_train_new.shape)
print("New test shape:", X_test_new.shape)
print("\nTraining class distribution:")
print(y_train_new.value_counts().sort_index())

print("\nTest class distribution:")
print(y_test_new.value_counts().sort_index())
```

```python
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.model_selection import cross_val_score

rf_cv_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=1
)

et_cv_model = ExtraTreesClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=1
)

rf_cv_scores = cross_val_score(
    rf_cv_model,
    X_train_new,
    y_train_new,
    cv=5,
    scoring="f1_macro",
    n_jobs=-1
)

et_cv_scores = cross_val_score(
    et_cv_model,
    X_train_new,
    y_train_new,
    cv=5,
    scoring="f1_macro",
    n_jobs=-1
)

print("Random Forest CV scores:", rf_cv_scores)
print(f"Random Forest mean F1: {rf_cv_scores.mean():.4f}")
print(f"Random Forest std: {rf_cv_scores.std():.4f}")

print("\nExtra Trees CV scores:", et_cv_scores)
print(f"Extra Trees mean F1: {et_cv_scores.mean():.4f}")
print(f"Extra Trees std: {et_cv_scores.std():.4f}")
```

```python
from sklearn.ensemble import ExtraTreesClassifier

final_model = ExtraTreesClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

final_model.fit(X_train_new, y_train_new)

print("Final model trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

final_pred = final_model.predict(X_test_new)

final_accuracy = accuracy_score(y_test_new, final_pred)
final_macro_f1 = f1_score(
    y_test_new, final_pred, average="macro"
)

print(f"Final Test Accuracy: {final_accuracy * 100:.2f}%")
print(f"Final Test Macro F1: {final_macro_f1:.4f}")

print("\nClassification Report:")
print(classification_report(y_test_new, final_pred))

cm = confusion_matrix(y_test_new, final_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=final_model.classes_
)

disp.plot(cmap="Blues", values_format="d")
plt.title("Final Extra Trees — Fresh Holdout")
plt.show()
```

```python
import joblib

joblib.dump(final_model, "forest_cover_model.pkl")

print("Final model saved successfully.")
```

## Final Model Selection and Conclusion

### Selected Model: Extra Trees Classifier

The Extra Trees Classifier was selected as the provisional final model for the Forest Cover Prediction project. It was chosen based on its cross-validation performance, fresh holdout results, and observed computational efficiency.

### Final Evaluation Results

- **Test accuracy:** 86.08%
- **Macro F1-score:** 0.8591
- **Correct predictions:** 2,603 out of 3,024
- **Cross-validation mean macro F1:** 0.8510 across five folds

### Why Extra Trees Was Selected

Extra Trees achieved a mean cross-validation macro F1-score of 0.8510, slightly higher than Random Forest's 0.8506. It also achieved 87.20% accuracy in the initial model comparison and trained faster in the benchmark run.

The difference between the two ensemble models was small, so the results do not establish that Extra Trees is definitively superior to Random Forest.

### Key Findings

1. Elevation was the most important feature in the initial Random Forest feature-importance analysis.
2. Classes 4 (Cottonwood/Willow) and 7 (Krummholz) were predicted particularly well.
3. Classes 1 (Spruce/Fir) and 2 (Lodgepole Pine) were more difficult to distinguish.
4. Tree-based ensemble methods outperformed the tested Decision Tree, KNN, and Logistic Regression baselines on the initial test split.

### Limitations

- The models were evaluated on one dataset, so performance on other geographic areas is unknown.
- The final holdout was newly partitioned, but the dataset had already been explored during earlier analysis and model comparisons.
- Feature importance represents model behavior, not causal relationships.
- Additional validation on independent data would strengthen confidence in production use.

### Conclusion

The project demonstrates a complete multiclass classification workflow, including exploratory data analysis, feature preparation, model comparison, cross-validation, final evaluation, and model persistence. The selected Extra Trees model achieved 86.08% accuracy on the fresh holdout. Further validation and deployment testing would be needed before relying on it in a real-world application.
